```python
import numpy as np
```

```python
def Phi(P, S, N):
    phi = 6.2 * P ** (0.18) * (S ** 2 / N) ** (0.1)
    return phi

def Delta(rho_s, rho_w):
    delta = (rho_s - rho_w) / rho_w
    return delta
```

```python
# input
P = 0.1
S = 2
N = 7000
Hs = 1.11
b = 2/3
xi = 1.93
psi = 2.5
alpha = 16.8 * ( np.pi) / 180
rho_s = 2082.4
rho_w = 1025
print(np.cos(alpha))
```

```python
Phii = Phi(P, S, N)
print(Phii)

Deltaa = Delta(rho_s, rho_w)
print(Deltaa)
```

```python
def D(Hs, delta, psi, phi, alpha, xi, b):
    DD = Hs * xi ** b / (delta * phi * psi * np.cos(alpha)) 
    return DD
```

```python
DD = D(Hs, Deltaa, psi, Phii, alpha, xi, b)
print(DD)
```
