```python
import numpy as np

def calculate_Rc(q, Hm0, tan_alpha, gamma_b, gamma_f, gamma_beta, gamma_v, h, width):
    g = 9.81  # gravity (m/s²)
    Rc = []
    xii = []
    tan_aa = []
    Ltott = []
    heightt = []

    for i in range(len(tan_alpha)):
        xi = tan_alpha[i] / np.sqrt(Hm0 / Lm)
        xii.append(xi)
        tan_aa.append(tan_alpha[i])

    # # # Calculate the argument inside the logarithm

        numerator = q * np.sqrt(tan_alpha[i])
        denominator = 0.026 * gamma_b * xi * np.sqrt(g * Hm0**3)
        log_argument = numerator / denominator
        # print(log_argument)
    
        if log_argument <= 0:
            raise ValueError("Logarithm argument must be positive.")
    
    # Compute the exponential expression
        else: 
            power_term = (-np.log(log_argument))**(1/1.3)
    
    # Final Rc computation
            Rcc = (xi * Hm0 * gamma_b * gamma_f * gamma_beta * gamma_v / 2.5) * power_term
            # print(Rcc)
            
            height = h +  Rcc
            Lin = height * 3.3
            Lout = height / tan_alpha[i]
            Ltot = Lout + width + Lin

            Rc.append(Rcc)
            heightt.append(height)
            Ltott.append(Ltot)
    return Rc, tan_aa, xii, heightt, Ltott
```

```python
# voor optimale berm
def optberm_Rc(q, Hm0, tan_alpha, gamma_f, gamma_beta, gamma_v, T, h, width):
    g = 9.81  # gravity (m/s²)
    Lm = g * T ** 2 / (2*np.pi)
    Rc = []
    xii = []
    tan_aa = []
    Bbermm = []
    gamma_bb = []
    Ltott = []
    heightt = []

    for i in range(len(tan_alpha)):
        xi = tan_alpha[i] / np.sqrt(Hm0 / Lm)
        xii.append(xi)
        tan_aa.append(tan_alpha[i])

        ## Calculate berm parameter:
        Bberm = min(4/3 * Hm0 * tan_alpha[i], 0.25 * Lm)
        Lberm = Bberm + 2 * Hm0 * tan_alpha[i]
        rb = Bberm / Lberm
        gamma_b = 1 - rb

        Bbermm.append(Bberm)
        gamma_bb.append(gamma_b)

    # # # Calculate the argument inside the logarithm

        numerator = q * np.sqrt(tan_alpha[i])
        denominator = 0.026 * gamma_b * xi * np.sqrt(g * Hm0**3)
        log_argument = numerator / denominator
        # print(log_argument)
    
        if log_argument <= 0:
            raise ValueError("Logarithm argument must be positive.")
    
    # Compute the exponential expression
        else: 
            power_term = (-np.log(log_argument))**(1/1.3)
    
    # Final Rc computation
            Rcc = (xi * Hm0 * gamma_b * gamma_f * gamma_beta * gamma_v / 2.5) * power_term
            # print(Rcc)
            
            height = h +  Rcc
            Lin = height * 3.3  
            Lout = height / tan_alpha[i] + Bberm
            Ltot = Lout + width + Lin
            Ltott.append(Ltot)
            heightt.append(height)

        # RC = ((xi * Hm0 * gamma_b * gamma_f * gamma_beta * gamma_v) / 2.5) * ((-np.log(q*np.sqrt(tan_alpha[i])/(0.026*gamma_b*xi*np.sqrt(g*(Hm0**3)))))**(1/1.3))
        Rc.append(Rcc)
    return Rc, tan_aa, xii, Bbermm, gamma_bb, heightt, Ltott
```

```python
def cost(tan_a, height, width, Bberm, h, C1, C2, C3, C4, C5, C6):
    # no berm taken into account
    Lin = height * 3.3  
    Lout = height / tan_a + Bberm
    Ltot = Lin + Lout + width
    vol_sand = 1/2 * Lin * (height -1) + width * (height - 1) + 1/2 * Lout * (height - 1) + Bberm * (h - 1)
    zijde_in = np.sqrt((Lin)**2 + height**2)
    zijde_out = np.sqrt(Lout ** 2 + height ** 2) + Bberm                   
    cost1 = C1 * Ltot
    cost2 = C2 * vol_sand
    cost3 = C3 * zijde_out
    cost4 = C4 * 0
    cost5 = C5 * (zijde_in + width)
    cost6 = C6 * (Ltot)
    tot = cost1 + cost2 + cost3 + cost4 + cost5 + cost6
    return tot, vol_sand

priceA = cost(0.303, 7.43, width, 0, h, 20, 50, 700, 350, 125, 100)
print(priceA[0])

priceB = cost(0.167, 6.4, width, 0, h, 20, 50, 700, 350, 125, 100)
print(priceB[0])

priceC = cost(0.303, 6.43, width, 1, h, 20, 50, 700, 350, 125, 100)
print(priceC[0])

priceD = cost(0.167, 5.88, width, 1, h, 20, 50, 700, 350, 125, 100)
print(priceD[0])
```
