```python
import numpy as np

def discharge1(
    g, Hm0, tan_a_list, gamma_b, gamma_f, gamma_beta, gamma_v,
    q_lim, Lm, Rc_list, width, DWL):
    tan_aa = []
    Rcc = []
    qq = []
    marginn = []
    xii = []
    
    # L_berm =


    for i in range(len(tan_a_list)):
        if tan_a_list[i] <= 0.303:
            xi = tan_a_list[i] / np.sqrt(Hm0 * Lm)
            for j in range(len(Rc_list)):

                # def overtopping_discharge(Hm0, Rc, tan_alpha, gamma_b, gamma_f, gamma_beta, gamma_v, xi_m_1_0):
                term1 = 0.026 / np.sqrt(tan_a_list[i])
                exponent = -((2.5 * Rc_list[j]) / (xi * Hm0 * gamma_b * gamma_f * gamma_beta * gamma_v))**1.3
                q_norm = term1 * gamma_b * xi * np.exp(exponent)
                q = q_norm * np.sqrt(g * Hm0**3)

                # q = (
                # np.sqrt(g * Hm0**3)
                # * 0.023 / np.sqrt(tan_a_list[i])
                # * gamma_b * xi
                # * np.exp(
                #     -(
                #         2.7 * Rc_list[j] / (xi * Hm0 * gamma_b * gamma_f * gamma_beta * gamma_v)
                #     ) ** 1.3
                # )
                # )
                margin = q_lim - q
                if margin > 0:
                    if margin < 0.01:
                        rest = q_lim - q
                        tan_aa.append(tan_a_list[i])
                        Rcc.append(Rc_list[j])
                        qq.append(q)
                        marginn.append(rest)
                        xii.append(xi)
                # else:
                    # print(f'tan_a = {tan_a:.3f}, Rc = {Rc:.2f} → q = {q:.4e} exceeds limit')

    return tan_aa, Rcc, qq, marginn, xii
```

```python
# Combination A
tan_a_list = np.linspace(0.1, 0.303, 50)
Rc_list = np.linspace(0, 10, 50)

A = discharge1(g, Hm0, tan_a_list, 1, gamma_f, gamma_beta, gamma_v, q_lim, Lm,  Rc_list , 5, 5.21 )
# print(D2[0])
# print(D2[1])
# print(D2[2])
# print(D2[3])

df = pd.DataFrame({
    'tan_a': A[0],
    'Rc': A[1],
    'q': A[2],
    'margin': A[3],
    'xi' : A[4]
})
print(df)
```

```python
# Costs
def cost(Ltot, height, width, C1, C2, C3, C4, C5, C6):
    vol_sand = (Ltot - width) * (height -1) + width * (height - 1)
    zijde = np.sqrt(((Ltot-width)/2)**2 + height**2)
    cost1 = C1 * Ltot
    cost2 = C2 * vol_sand
    cost3 = C3 * zijde
    cost4 = C4 * 0
    cost5 = C5 * (zijde + width)
    cost6 = C6 * zijde
    tot = cost1 + cost2 + cost3 + cost4 + cost5 + cost6
    return tot, vol_sand, zijde

price = cost(39.39, 5.21, 5, 20, 50, 700, 350, 125, 100)
print(price)
```
